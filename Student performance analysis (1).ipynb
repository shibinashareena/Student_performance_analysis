### HANDLING MISSING VALUES

#### Missing values are replaced using the mode of each column. Mode is the most frequently occurring value in a column. This method helps to fill incomplete records without deleting rows from the dataset.

```python
df = df.fillna(df.mode().iloc[0])
```

```python
df.isnull().sum()
```

#### Missing values were successfully filled using the mode of each column. After filling, the dataset was checked again, and no missing values were found.

### CREATING DERIVED COLUMNS

#### A derived column is a new column created using existing data. Here, we created a performance_level column based on students's exam scores.

```python
df["Performance_Level"] = "Needs Improvement"
```

```python
df.loc[df["Exam_Score"] >= 50, "Performance_Level"] = "Average"
df.loc[df["Exam_Score"] >= 75, "Performance_Level"] = "Good"
```

```python
df.head()
```

#### A new column named Performance_Level is created based on students's exam scores. Students scoring 75 or above are classied as Good, and the others as Average.

### FILTERING AND AGGREGATING

#### Filtering and aggregating data helps us summarize the dataset. Here, we count the number of students in each performance category, such as Average and Good. This helps us understand the distribution of student performance.

```python
df["Performance_Level"].value_counts()
```

#### The analysis shows that 6,483 students are classified as Average and 124 students are classified as Good based on the selected exam score categories. Most students fall under the Average category.

### UNIVARIATE ANALYSIS

### BIVARIATE ANALYSIS

#### Bivariate analysis is used to study the relationship between two variables. Here, we analyze the relationship between students' study hours and exam scores to understand whether studying more hours is associated with better performance.

### MULTIVARIATE ANALYSIS

#### Multivariate analysis is used to study three or more variables together. Here, we analyze the relationship between study hours, attendance, and exam scores.

```python
df[["Hours_Studied","Attendance","Exam_Score"]].corr()
```

#### The correlation analysis shows the relationships between study hours, attendance, and exam scores.
#### Hours_Studied and Exam_Score: 0.445 – Moderate positive correlation.
#### Attendance and Exam_Score: 0.581 – Moderate positive correlation.
#### Hours_Studied and Attendance: -0.010 – Almost no correlation.
#### Conclusion: Attendance has a stronger positive relationship with exam scores than study hours in this dataset.

### DATA VISUALIZATION

#### HISTOGRAM OF EXAM SCORES

#### A histogram is used to show the distribution of student's exam scores. It helps us understand how scores are spread across the dataset.

```python
plt.hist(df["Exam_Score"])
plt.title("Distribution of Exam Scores")
plt.xlabel("Exam Score")
plt.ylabel("Number of Students")
plt.show()
```

#### The histogram shows the distribution of student's exam scores. Most students scored between 60 and 75, while fewer students scored above 75. This indicates that most students have scores concentrated around the average range

#### BAR CHART OF AVERAGE EXAM SCORES BY GENDER

#### A bar chart is used to compare the average exam scores of male and female students.

```python
df.groupby("Gender")["Exam_Score"].mean().plot(kind="bar")
plt.title("Average Exam Score by Gender")
plt.xlabel("Gender")
plt.ylabel("Average Exam Score")
plt.show()
```

```python
df.groupby("Gender")["Exam_Score"].mean()
```

#### Female students have a slightly higher average exam score than male students. However, the difference is very small, so both groups have almost equal performance.

#### SCATTER PLOT - STUDY HOURS VS EXAM SCORE

```python
plt.scatter(df["Hours_Studied"],df["Exam_Score"])
plt.title("Study Hours vs Exam Score")
plt.xlabel("Hours Studied")
plt.ylabel("Exam Score")
plt.show()
```

#### The scatter plot shows that exam scores generally increase slightly as study hours increase.

#### LINE CHART - ATTENDANCE VS AVERAGE SCORES

#### This line chart shows the relationship between students' attendance and their average exam scores.

```python
df.groupby("Attendance")["Exam_Score"].mean().plot(kind="line")

plt.title("Attendance vs Average Exam Scores")
plt.xlabel("Attendance")
plt.ylabel("Average Exam Score")
plt.show()
```

#### The line chart shows that students with higher attendance generally have higher average exam scores. This indicates a positive relationship between attendance and academic performance.

#### PIE CHART - SCHOOL TYPE DISTRIBUTION

#### This pie chart shows the distribution of students across different school types.

```python
df["School_Type"].value_counts().plot(kind="pie")
plt.title("School Type Distribution")
plt.show()
```

#### The pie chart indicates that most students belong to public schools, while fewer students belong to private schools. This shows that the number of students in public schools is higher than in private schools.

#### BOX PLOT SCORES BY MOTIVATION LEVEL

#### The box plot shows the distribution of student's exam scores across different motivation levels. It helps compare the scores and understand how motivation levels affect student's academic performance.

```python
plt.figure(figsize=(8,5))
sns.boxplot(data=df, x="Motivation_Level", y="Exam_Score")
plt.title("Scores by Motivation Level")
plt.xlabel("Motivation Level")
plt.ylabel("Exam Score")
plt.show()
```

#### The box plot displays the exam score distribution for Low, Medium, and High motivation levels. It shows that the median scores are similar across all three groups, with some variations and outliers in each group.

#### HEATMAP - NUMERICAL CORRELATION

#### This heatmap shows the correlation between numerical variables in the student performance dataset. It helps identify positive and negative relationships between different numerical features.

#### The heatmap displays the correlation between numerical variables in the student performance dataset. It shows that attendance and hours studied have a positive relationship with exam scores, while other variables show different levels of correlation.

```python
plt.figure(figsize=(10,6))
sns.heatmap(df.corr(numeric_only=True))
plt.title("Numerical Correlation Heatmap")
plt.show()
```

#### The heatmap displays the correlation between numerical variables in the student performance dataset. It shows that attendance and hours studied have a positive relationship with exam scores, while other variables show different levels of correlation.

#### COUNT PLOT - PARENTAL INVOLVEMENT

#### This count plot shows the number of students based on different levels of parental involvement. It helps understand how parental involvement is distributed among students.

```python
sns.countplot(data=df, x="Parental_Involvement")
plt.title("Parental Involvement")
plt.xlabel("Parental Involvement")
plt.ylabel("Count")
plt.show()
```

#### The count plot shows that most students have a medium level of parental involvement, followed by high and low levels. This indicates that medium parental involvement is the most common among students.

#### VIOLIN PLOT - SCORES BY RESOURCE ACCESS

#### This violin plot shows the distribution of students' exam scores based on different levels of resource access. It helps compare the score distribution across resource access levels.

```python
sns.violinplot(data=df, x="Access_to_Resources", y="Exam_Score")
plt.title("Scores by Resource Access")
plt.show()
```

#### The violin plot shows that students with high resource access generally have slightly higher exam scores than those with medium and low resource access. It also displays the distribution and variation of scores across all three groups.

#### PAIR PLOT - RELATIONSHIP BETWEEN NUMERICAL VARIABLES

#### This pair plot shows the relationships between different numerical variables in the student performance dataset. It helps identify patterns and relationships between variables.

```python
sns.pairplot(df[["Hours_Studied", "Attendance", "Previous_Scores", "Exam_Score"]].sample(200))
plt.show()
```

#### The pair plot shows the relationships between numerical variables in the student performance dataset. It helps identify patterns and correlations between different variables.
